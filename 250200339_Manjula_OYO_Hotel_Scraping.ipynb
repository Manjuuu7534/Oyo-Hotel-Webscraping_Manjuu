```python
response = requests.get(page_url, headers={"User-Agent": "Mozilla/5.0"})

print("Status Code:", response.status_code)
print("Page Length:", len(response.text))
```

```python
soup = BeautifulSoup(response.content, "html.parser")
print(soup.title)
```

```python
hotels = soup.find_all(
    'div',
    class_="oyo-cell--12-col oyo-cell--8-col-tablet oyo-cell--4-col-phone"
)

print("Total Hotels:", len(hotels))
```

```python
hotel_list = []

for hotel in hotels:
    name = hotel.h3["title"]

    address_tag = hotel.find('span', class_="u-line--clamp-2")
    rating_tag = hotel.find('span', class_="hotelRating__rating")
    num_rating_tag = hotel.find('span', class_="hotelRating__ratingSummary")
    price_tag = hotel.find('span', class_="listingPrice__finalPrice")

    address = address_tag["title"] if address_tag else "Not Available"
    star_rating = rating_tag.text.strip() if rating_tag else "Not Available"
    num_rating = num_rating_tag.text.strip() if num_rating_tag else "Not Available"
    price = price_tag.text.strip() if price_tag else "Not Available"

    hotel_list.append({
        "Title": name,
        "Address": address,
        "Star_Rating": star_rating,
        "Rating_Count": num_rating,
        "Price": price
    })

hotel_list
```

```python
df = pd.DataFrame(hotel_list)
df
```

**Question 1: Which hotel has the highest rating count, and what is its star rating and price?**

```python

from sqlalchemy import text

query = """
SELECT "Title", "Star_Rating", "Rating_Count", "Price"
FROM public.oyo_hotels
ORDER BY NULLIF(
    REGEXP_REPLACE("Rating_Count", '[^0-9]', '', 'g'), ''
)::INTEGER DESC NULLS LAST
LIMIT 1;
"""

with engine.connect() as connection:
    result = connection.execute(text(query))
    for row in result:
        print(row)
```

**What is the average price of all scraped hotels?**

```python
query = """
SELECT
    ROUND(AVG(
        CAST(REGEXP_REPLACE("Price", '[^0-9]', '', 'g') AS NUMERIC)
    ), 2) AS average_price
FROM public.oyo_hotels;
"""

with engine.connect() as connection:
    result = connection.execute(text(query))
    for row in result:
        print("Average Price: ₹", row[0])
```

**Which hotel has the highest star rating, and what is its price?**

```python

query = """
SELECT "Title", "Star_Rating", "Price"
FROM public.oyo_hotels
WHERE "Star_Rating" ~ '^[0-9]+(\\.[0-9]+)?$'
ORDER BY CAST("Star_Rating" AS NUMERIC) DESC
LIMIT 1;
"""

with engine.connect() as connection:
    result = connection.execute(text(query))
    for row in result:
        print(row)
```

**What is the average star rating across all hotels?**

```python

query = """
SELECT
    ROUND(AVG(CAST("Star_Rating" AS NUMERIC)), 2)
    AS average_star_rating
FROM public.oyo_hotels
WHERE "Star_Rating" ~ '^[0-9]+(\\.[0-9]+)?$';
"""

with engine.connect() as connection:
    result = connection.execute(text(query))
    for row in result:
        print("Average Star Rating:", row[0])
```
